#### The columns that were deciphered.

- 's1aq1'- gender	
- 's1aq2'- role in the household.
- 's2aq1' - Education as whether peson has attended school 
- 's2aq3' - highest level of education
- 's4aq40a' - occupation
- 's4aq42' - occupation type (registration)
- 's4aq44' - occupation place
- 's1aq4y'- age
- 's1aq9a' - tribe
- 's1bq5a' - region of stay
- 's1aq5' - marital status


```python
narrow_microdata.info()
```

##### Renaming deciphered columns to reflect actual context meaning.
- Also included original codes to for easy tracking

```python
narrow_microdata = narrow_microdata.rename(columns={'s1aq1': 's1aq1_gender','s1aq2': 's1aq2_ role_in_household','s2aq1': 's2aq1_education',
                               's2aq3': 's2aq3_highest_level_of_education', 's4aq40a':'s4aq40a_occupation','s4aq42': 's4aq42_occupation_type_registration',
                                's4aq44':'s4aq44_occupation_place','s1aq4y':'s1aq4y_age','s1aq9a':'s1aq9a_tribe','s1bq5a':'s1bq5a_region of stay',
                                's1aq5':'s1aq5_marital_status'})

narrow_microdata.head()
```

##### Limiting the occupation to only farmer

```python
farmer_data =narrow_microdata[narrow_microdata["s4aq40a_occupation"].str.contains(" farmer ", case=False, na=False)]
farmer_data.head()
```

```python
farmer_data.info()
```

##### Limiting ages to 18 and above

```python
farmer_data_above =farmer_data[farmer_data['s1aq4y_age']>17]
farmer_data_above.head()
```

```python
# Checking for unique value counts of occupation
farmer_data_above['s4aq40a_occupation'].value_counts()
```

```python
crop_type = list(farmer_data_above['s4aq40a_occupation'].unique())
crop_type #Checking for irrelevant roles
```

Words removed : labourer, livestock farmer, poultry farmer, fish farmer, worker

```python

farmer_data_above = farmer_data_above[~farmer_data_above["s4aq40a_occupation"].str.contains("worker", case=False, na=False)] # repeated this step until all the words mentioned above were removed
crop_type = list(farmer_data_above['s4aq40a_occupation'].unique())
```
